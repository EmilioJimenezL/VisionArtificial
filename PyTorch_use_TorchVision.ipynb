```python
import cv2
from ultralytics import YOLO
```

```python
model = YOLO("yolov8n.pt")
```

```python
capture = cv2.VideoCapture(0)
```

```python
while True:
    ret, frame = capture.read()
    results = model(frame)
    visual = results[0].plot
    cv2.imshow(visual)
    if cv2.waitKey(1) & 0xFF == ord('q'):
        break

capture.release
cv2.destroyAllWindows
```
